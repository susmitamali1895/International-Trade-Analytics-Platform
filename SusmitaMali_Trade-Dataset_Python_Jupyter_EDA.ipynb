```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Data.csv")
df
```

```python
df.columns= ["Year_Month", "Trade_Type","Partner_Country_Code",
"Product_Category_Code", "HS_Product_Code", "Flag","Quantity", "Trade_Value"]
```

```python
print(df.columns)
```

```python
# Clean column names 
df.columns = (df.columns 
             .str.strip()
             .str.lower()
             )
df.columns
```

```python
# Value counts 
categorical_cols = [
    'year_month',
    'trade_type',
    'partner_country_code',
    'product_category_code',
    'hs_product_code','flag'
]
for col in categorical_cols:
    print(f"\nValue counts for {col}")
    print(df[col].value_counts())
```

```python
# Find null values 
print("\nNull Value Count:")
print(df.isnull().sum())
```

```python
df = df.fillna(df.mean(numeric_only=True))
print("\nMissing values after:")
print(df.isnull().sum())
```

```python
# Duplicate Rows 
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Top 10 Product Categories by Trade Value
product = (df.groupby('product_category_code')['trade_value'].sum().sort_values(ascending=False).head(10))
plt.figure(figsize=(10,5))
product.plot(kind='bar')
plt.title("Top 10 Product Categories by Trade value" )
plt.xlabel("Product Category")
plt.ylabel("Trade Value")
plt.xticks(rotation=45)
plt.show()
```

```python
# Top 10 Partner countries by Trade Value 
top_country = (
    df.groupby('partner_country_code')['trade_value'].sum().sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))
top_country.plot(kind='bar', color='purple')
plt.title('Top 10 Partner Countries by Trade Value')
plt.xlabel('Partner Country Code')
plt.ylabel('Trade Value')
plt.xticks(rotation=45)
plt.show()
```

```python
# Trade Value vs HS Product code
top_hs = (df.groupby('hs_product_code')['trade_value'].sum().sort_values(ascending=False).head(10))

plt.figure(figsize=(10,5))
top_hs.plot(kind='barh',color='pink')
plt.title('Top 10 HS Product Codes by Trade Value')
plt.xlabel('Trade Value')
plt.ylabel('HS Product Code')
plt.xticks(rotation=45)
plt.show()
```

```python
# Top Product Categories by Quantity 
top_prod = (
    df.groupby('product_category_code')['quantity']
      .sum().sort_values().tail(8))
# convert to millions
top_prod_m = top_prod / 1_000_000

top_prod_m.plot(kind='barh', figsize=(8,5))
plt.xlabel('Total Quantity (in Millions)')
plt.title('Top Product Categories by Quantity')
plt.tight_layout()
plt.show()
```

```python
# Quatity vs Trade Value 
plt.figure(figsize=(8,5))
plt.scatter(df['quantity'], df['trade_value'], alpha=0.5)
plt.xlabel('Quantity')
plt.ylabel('Trade Value')
plt.title('Quantity vs Trade Value Relationship')
plt.show()
```

```python
# Trade Value by Trade Type 
trade_type_value = (
    df.groupby('trade_type')['trade_value'].sum())

plt.figure(figsize=(6,4))
trade_type_value.plot(kind='barh',color='blue')
plt.title('Trade Value by Trade Type')
plt.xlabel('Trade Value')
plt.ylabel('Trade Type')
plt.xticks(rotation=0)
plt.show()
```

```python
# Import vs Export Distribution 
trade_type_map = {1: 'Import', 2: 'Export'}
df_distribution = df.groupby('trade_type')['quantity'].sum()

plt.pie(df_distribution, labels=df_distribution.index, autopct='%1.1f%%', startangle=90)
plt.title('Import vs Export Quantity Share')
plt.show()
```

```python
# Top HS code by Quantity 
top_hs_qty = (
    df.groupby('hs_product_code')['quantity'].sum().sort_values(ascending=False).head(10)
)
plt.figure(figsize=(10,5))
top_hs_qty.plot(kind='bar', color ='red')
plt.title('Top 10 HS Product Codes by Quantity')
plt.xlabel('HS Product Code')
plt.ylabel('Quantity')
plt.xticks(rotation=45)
plt.show()
```

```python
# Number of Transaction Count by Trade Type 
trade_type_map = {1: 'Import', 2: 'Export'}
trade_count = df.groupby('trade_type').size()

# convert to millions
trade_count_m = trade_count / 1_000_000

plt.figure(figsize=(6,4))
trade_count_m.plot(kind='bar')

plt.xlabel('Trade Type')
plt.ylabel('Number of Transactions (in Millions)')
plt.title('Trade Transactions by Trade Type')

plt.tight_layout()
plt.show()
```

```python
monthly_transaction = df.groupby('year_month').size().sort_index()

plt.figure(figsize=(10,5))
plt.plot(monthly_transaction.index, monthly_transaction.values, color='brown')
plt.xlabel('Year-Month')
plt.ylabel('Number of Transactions')
plt.title('Monthly Trade Activity Trend')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
